```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.shape
df.columns
df.info()
df.describe()
df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage.sort_values(ascending=False)
```

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
df.dtypes
```

```python
if 'publishedAt' in df.columns:
    df['publishedAt'] = pd.to_datetime(df['publishedAt'], errors='coerce')
df.head()    
```

```python
if 'category_id' in df.columns:
    print(df['category_id'].value_counts())
```

```python
if 'category_id' in df.columns:
    plt.figure(figsize=(10,5))
    df['category_id'].value_counts().head(10).plot(kind='bar')
    plt.title('Top YouTube Categories')
    plt.xlabel('Category_id')
    plt.ylabel('Number of Videos')
    plt.xticks(rotation=45)
    plt.show()
if 'view_count' in df.columns:
    print(df['view_count'].describe())
```

```python
top10 = df.sort_values(by='views', ascending=False).head(10)

top10[['title', 'views']]
```

```python
top10 = df.sort_values('views', ascending=False).head(10).copy()

# Shorten long titles
top10['short_title'] = top10['title'].astype(str).apply(
    lambda x: x[:50] + '...' if len(x) > 50 else x
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top10,
    x='views',
    y='short_title'
)

plt.title('Top 10 Most Viewed YouTube Videos', fontsize=16)
plt.xlabel('Views')
plt.ylabel('Video Title')

plt.tight_layout()
plt.show()
```

```python
top10 = df.sort_values('views', ascending=False).head(10).copy()

top10['views_millions'] = top10['views'] / 1_000_000

top10['short_title'] = top10['title'].astype(str).apply(
    lambda x: x[:45] + '...' if len(x) > 45 else x
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top10,
    x='views_millions',
    y='short_title'
)

plt.title('Top 10 Most Viewed YouTube Videos', fontsize=16)
plt.xlabel('Views (Millions)')
plt.ylabel('Video Title')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='views',
    y='likes'
)

plt.title('Views vs Likes', fontsize=16)
plt.xlabel('Views')
plt.ylabel('Likes')

plt.tight_layout()
plt.show()
```
